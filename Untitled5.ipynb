# PART A — Data Exploration and Cleaning

## 1. Import the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Machine learning libraries
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression

# Evaluation metrics
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

# Save/load model
import joblib
```

## 2. Load the dataset

```python
df = pd.read_csv('/content/house_price_prediction_dataset.csv')

print('First five records:')
display(df.head())

print('\nLast five records:')
display(df.tail())
```

## 3. Check the shape

```python
print('Number of rows:', df.shape[0])
print('Number of columns:', df.shape[1])
```

## 4. Check the data types

```python
display(df.dtypes)
```

## 5. Display dataset information and summary statistics

```python
print('Dataset Information:')
display(df.info())

print('\nSummary Statistics:')
display(df.describe())

print('\nSummary Statistics (including categorical):')
display(df.describe(include='all'))
```

## 6. Check missing values

```python
print('Missing values count per column:')
display(df.isnull().sum())

print('\nMissing values percentage per column:')
missing = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': df.isnull().mean() * 100
})
display(missing)
```

## 7. Check duplicate records

```python
print('Number of duplicate rows:', df.duplicated().sum())

print('\nDuplicate rows:')
display(df[df.duplicated(keep=False)])
```

## 8. Remove duplicates

```python
df = df.drop_duplicates()

print('New shape:', df.shape)
```

## 9. Check missing target values

```python
print('Missing target values before removal:', df['House_Price_Million_RWF'].isnull().sum())

df = df.dropna(subset=['House_Price_Million_RWF'])

print('Missing target values after removal:', df['House_Price_Million_RWF'].isnull().sum())
print('New shape after removing rows with missing target values:', df.shape)
```

## 10. Detect outliers using the IQR method

```python
numeric_columns = [
    'Area_m2',
    'Bedrooms',
    'Bathrooms',
    'House_Age_Years',
    'Distance_to_City_km',
    'Parking_Spaces',
    'House_Price_Million_RWF'
]

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f'Column: {column}')
    print('  Lower bound:', lower_bound)
    print('  Upper bound:', upper_bound)
    print('  Number of outliers:', len(outliers))
    print()
```

## 11. Visualize outliers with boxplots

```python
plt.figure(figsize=(12, 8))

df[numeric_columns].boxplot()

plt.xticks(rotation=45)
plt.title('Boxplots of Numeric Variables')
plt.grid(False)
plt.show()

plt.figure(figsize=(8, 5))

sns.boxplot(
    x=df['House_Price_Million_RWF']
)

plt.title('House Price Boxplot')
plt.xlabel('Price (Million RWF)')
plt.show()
```

## 12. Cap extreme values

```python
for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(
        lower_bound,
        upper_bound
    )
print('Extreme values capped for numeric columns.')
```

## 13. Save the cleaned dataset

```python
df.to_csv(
    'cleaned_house_price_prediction_dataset.csv',
    index=False
)
print('Cleaned dataset saved as cleaned_house_price_prediction_dataset.csv')
```

## 14. Visualization 1 — Price distribution

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df['House_Price_Million_RWF'],
    kde=True
)

plt.title('Distribution of House Prices')
plt.xlabel('House Price (Million RWF)')
plt.ylabel('Frequency')

plt.show()
```

## 15. Visualization 2 — Area vs Price

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Area_m2',
    y='House_Price_Million_RWF'
)

plt.title('Area vs House Price')
plt.xlabel('Area (m²)')
plt.ylabel('Price (Million RWF)')

plt.show()
```

## 16. Visualization 3 — Price by neighborhood

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Neighborhood',
    y='House_Price_Million_RWF'
)

plt.xticks(rotation=30)

plt.title('House Price by Neighborhood')
plt.xlabel('Neighborhood')
plt.ylabel('Price (Million RWF)')

plt.show()
```

## 17. Visualization 4 — Correlation heatmap

```python
plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.show()
```

```python
X = df.drop(
    columns=[
        'House_ID',
        'House_Price_Million_RWF'
    ]
)

y = df['House_Price_Million_RWF']

print('Shape of X:', X.shape)
print('Shape of y:', y.shape)
```

## 19. Separate numeric and categorical variables

```python
numeric_features = [
    'Area_m2',
    'Bedrooms',
    'Bathrooms',
    'House_Age_Years',
    'Distance_to_City_km',
    'Parking_Spaces'
]

categorical_features = [
    'Neighborhood'
]

print('Numeric features:', numeric_features)
print('Categorical features:', categorical_features)
```

## 20. Split the data into training and testing

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print('Training data (X_train):', X_train.shape)
print('Testing data (X_test):', X_test.shape)
print('Training target (y_train):', y_train.shape)
print('Testing target (y_test):', y_test.shape)
```

## 21. Handle missing numeric values

```python
numeric_transformer = Pipeline(
    steps=[
        (
            'imputer',
            SimpleImputer(strategy='median')
        )
    ]
)
print('Numeric transformer (median imputation) created.')
```

## 22. Handle the Neighborhood variable (One-Hot Encoding)

```python
categorical_transformer = Pipeline(
    steps=[
        (
            'imputer',
            SimpleImputer(
                strategy='most_frequent'
            )
        ),
        (
            'onehot',
            OneHotEncoder(
                drop='first',
                handle_unknown='ignore'
            )
        )
    ]
)
print('Categorical transformer (most frequent imputation + One-Hot Encoding) created.')
```

## 23. Combine preprocessing

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            numeric_transformer,
            numeric_features
        ),
        (
            'cat',
            categorical_transformer,
            categorical_features
        )
    ]
)
print('ColumnTransformer for preprocessing created.')
```

## 24. Create the Linear Regression pipeline

```python
model = Pipeline(
    steps=[
        (
            'preprocessor',
            preprocessor
        ),
        (
            'regressor',
            LinearRegression()
        )
    ]
)
print('Linear Regression model pipeline created.')
```

## 25. Train the model

```python
model.fit(
    X_train,
    y_train
)
print('Model training complete.')
```

# PART C — Evaluation

## 26. Make predictions

```python
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print('Predictions made for training and testing sets.')
```

## 27. Calculate R²

```python
train_r2 = r2_score(
    y_train,
    y_train_pred
)

test_r2 = r2_score(
    y_test,
    y_test_pred
)

print(f'Training R²: {train_r2:.4f}')
print(f'Testing R²: {test_r2:.4f}')
```

## 28. Calculate MAE

```python
train_mae = mean_absolute_error(
    y_train,
    y_train_pred
)

test_mae = mean_absolute_error(
    y_test,
    y_test_pred
)

print(f'Training MAE: {train_mae:.2f}')
print(f'Testing MAE: {test_mae:.2f}')
```

## 29. Calculate RMSE

```python
train_rmse = np.sqrt(
    mean_squared_error(
        y_train,
        y_train_pred
    )
)

test_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_test_pred
    )
)

print(f'Training RMSE: {train_rmse:.2f}')
print(f'Testing RMSE: {test_rmse:.2f}')
```

## 30. Calculate adjusted R²

```python
n_train = len(y_train)

p = model.named_steps[
    'preprocessor'
].transform(X_train).shape[1]

adjusted_r2_train = (
    1 -
    (1 - train_r2) *
    (n_train - 1) /
    (n_train - p - 1)
)

n_test = len(y_test)

# Recalculate p for the test set if needed, or assume it's the same based on preprocessor fit on train
# For consistent comparison, p is derived from the preprocessor fitted on X_train
adjusted_r2_test = (
    1 -
    (1 - test_r2) *
    (n_test - 1) /
    (n_test - p - 1)
)

print(f'Training Adjusted R²: {adjusted_r2_train:.4f}')
print(f'Testing Adjusted R²: {adjusted_r2_test:.4f}')
```

## 31. Create the evaluation table

```python
results = pd.DataFrame({
    'Dataset': [
        'Training',
        'Testing'
    ],
    'R2': [
        train_r2,
        test_r2
    ],
    'Adjusted_R2': [
        adjusted_r2_train,
        adjusted_r2_test
    ],
    'MAE': [
        train_mae,
        test_mae
    ],
    'RMSE': [
        train_rmse,
        test_rmse
    ]
})

display(results)
```

## 32. Get the coefficients

```python
preprocessor_fitted = model.named_steps[
    'preprocessor'
]

feature_names = (
    preprocessor_fitted
    .get_feature_names_out()
)

coefficients = (
    model.named_steps[
        'regressor'
    ].coef_
)

coefficient_table = pd.DataFrame({
    'Variable': feature_names,
    'Coefficient': coefficients
})

display(coefficient_table)

intercept = (
    model.named_steps[
        'regressor'
    ].intercept_
)

print(f'Intercept: {intercept:.4f}')
```

# PART E — Streamlit Application

First, we will create the `app.py` file with the Streamlit application code. This file will be saved in your current Colab environment.

```python
streamlit_app_code = '''
import streamlit as st
import pandas as pd
import joblib


# Page configuration
st.set_page_config(
    page_title="Rwanda House Price Predictor",
    page_icon="🏠"
)


# Load model
@st.cache_resource
def load_model():

    model = joblib.load(
        "house_price_model.sav"
    )

    return model


model = load_model()


# Title
st.title("🏠 Rwanda House Price Predictor")

st.write(
    """
    This application predicts the estimated market price
    of a house in million Rwandan francs (RWF) using
    multiple linear regression.
    """
)


# User inputs

area = st.number_input(
    "Area (m²)",
    min_value=20.0,
    max_value=300.0,
    value=120.0,
    step=1.0
)


bedrooms = st.number_input(
    "Number of Bedrooms",
    min_value=1,
    max_value=6,
    value=3,
    step=1
)


bathrooms = st.number_input(
    "Number of Bathrooms",
    min_value=1,
    max_value=5,
    value=2,
    step=1
)


age = st.number_input(
    "House Age (Years)",
    min_value=0.0,
    max_value=35.0,
    value=5.0,
    step=0.5
)


distance = st.number_input(
    "Distance to City Centre (km)",
    min_value=0.0,
    max_value=25.0,
    value=4.0,
    step=0.1
)


parking = st.number_input(
    "Parking Spaces",
    min_value=0,
    max_value=3,
    value=1,
    step=1
)


neighborhood = st.selectbox(
    "Neighborhood",
    [
        "Gasabo",
        "Huye",
        "Kicukiro",
        "Kigali City",
        "Musanze",
        "Nyarugenge"
    ]
)


# Prediction button

if st.button("Predict"):

    input_data = pd.DataFrame({

        "Area_m2": [area],

        "Bedrooms": [bedrooms],

        "Bathrooms": [bathrooms],

        "House_Age_Years": [age],

        "Distance_to_City_km": [distance],

        "Parking_Spaces": [parking],

        "Neighborhood": [neighborhood]
    })


    prediction = model.predict(
        input_data
    )


    predicted_price = prediction[0]


    st.success(
        f"Estimated House Price: "
        f"{predicted_price:.2f} million RWF"
    )
'''

with open('app.py', 'w') as f:
    f.write(streamlit_app_code)

print('app.py created successfully.')
```
